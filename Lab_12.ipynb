# MATRIX CHAIN MULTIPLICATION Dynamic Programming with table filling.
Problem Description:
Find minimum number of scalar multiplications needed to multiply a chain of matrices..

#Theory
Cost of multiplying two matrices:
If A is p×q and B is q×r, cost = p × q × r

Dynamic Programming idea:
Break problem into smaller subproblems.
Store results in a table to avoid recomputation.

m[i][j] = minimum cost to multiply Ai to Aj

## Algorithm
1. Take array p[] of size n
2. Create table m[n][n]
3. Set m[i][i] = 0
4. Fill table diagonally
5. For each chain length L:
    Compute m[i][j]
    Try all possible k between i and j
6. Choose minimum cost
7. Return m[1][n-1]

```python
import sys

def matrix_chain_multiplication(p):
    n = len(p)

    # Create DP table
    m = [[0 for _ in range(n)] for _ in range(n)]

    # L is chain length
    for L in range(2, n):
        for i in range(1, n - L + 1):
            j = i + L - 1
            m[i][j] = sys.maxsize

            for k in range(i, j):
                cost = (m[i][k] +
                        m[k+1][j] +
                        p[i-1] * p[k] * p[j])

                if cost < m[i][j]:
                    m[i][j] = cost

    return m
```

```python
# Input dimensions
p = [10, 30, 5, 60]

table = matrix_chain_multiplication(p)

print("Minimum multiplication cost:", table[1][len(p)-1])

print("\nDP Table:")
for row in table[1:]:
    print(row[1:])
```

```python
import sys

def matrix_chain_order(p):
    n = len(p) - 1

    # m[i][j] = minimum cost
    m = [[0]*n for _ in range(n)]

    # s[i][j] = split index
    s = [[0]*n for _ in range(n)]

    for L in range(2, n+1):  # chain length
        for i in range(n - L + 1):
            j = i + L - 1
            m[i][j] = sys.maxsize

            for k in range(i, j):
                cost = m[i][k] + m[k+1][j] + p[i]*p[k+1]*p[j+1]

                if cost < m[i][j]:
                    m[i][j] = cost
                    s[i][j] = k

    return m, s
```

```python
def print_optimal_parens(s, i, j):
    if i == j:
        print(f"A{i+1}", end="")
    else:
        print("(", end="")
        print_optimal_parens(s, i, s[i][j])
        print_optimal_parens(s, s[i][j] + 1, j)
        print(")", end="")
```

```python
def naive_cost(p):
    cost = 0
    rows = p[0]
    cols = p[1]

    for i in range(1, len(p)-1):
        cost += rows * cols * p[i+1]
        cols = p[i+1]

    return cost
```

```python
p = [5, 10, 3, 12, 5, 50, 6]

m, s = matrix_chain_order(p)

# 1. Minimum Cost
print("Minimum number of multiplications:", m[0][len(p)-2])

# 2. Optimal Parenthesization
print("Optimal Parenthesization: ", end="")
print_optimal_parens(s, 0, len(p)-2)
print()

# 3. Comparison
naive = naive_cost(p)
optimal = m[0][len(p)-2]

print("Naive cost:", naive)
print("Optimal cost:", optimal)
print("Multiplications saved:", naive - optimal)
```
